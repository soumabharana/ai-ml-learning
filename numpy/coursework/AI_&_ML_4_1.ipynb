```python
a = [3, 5, 6]
b = [5, 6, 1]

# c = [8,11,7]
c = []

n = len(a)
for i in range(n):
    c.append(a[i] + b[i])

print(c)
```

```python
an = np.array(a)
bn = np.array(b)
print(an + bn)
```

```python
print(an-bn)
```
