```python
# Pass the prompt string directly inside input()
a = int(input("Enter 1st Number: "))
b = int(input("Enter 2nd Number: "))

c = a + b

# String formatting (f-string) to print everything in one line
print(f"Sum of 2 Numbers is: {c}")
```

```python
# Taking two integer inputs
a = int(input())
b = int(input())

# Arithmetic Operations
print("Addition: ", a + b)
print("Subtraction: ", a - b)
print("Multiplication: ", a * b)
print("Division: ", a / b)
print("Floored Division: ", a // b)
print("Modulus: ", a % b)
print("Power: ", a ** b)
```
