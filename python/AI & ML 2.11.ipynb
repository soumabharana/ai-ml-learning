#Map

```python
def cube(lst):
  cb = []
  for elem in lst:
    cb.append(elem**3)

  return cb
```

```python
cube([1,2,3,4])
```

```python
lst = [4,6,2]
cube = map(lambda x: x**3,lst)
print(cube)
```

```python
lst = [4,6,2]
cube = list(map(lambda x: x**3,lst))
print(cube)
```

```python
cube = list(map(lambda x: x**3,[4,6,8]))
print(cube)
```

```python
print(list(map(lambda x: x**3,[4,6,8])))
```

```python
def add(lst):
  ad=[]
  for elem in lst:
    ad.append(elem+5)
  return ad
```

```python
def upper(strings):
  str =[]
  for elem in strings:
    str.append(elem.upper())
  return str
```

```python
upper(["Ajay", "Neha","Akash","Beena"])
```

```python
strings = ["Ajay", "Neha","Akash","Beena"]
upper = list(map(lambda str: str.upper(),strings))
print(upper)
```

```python
print(list(map(lambda str: str.upper(),["Ajay", "Neha","Akash","Beena"])))
```

```python
def gst(prices):
  tax=[]
  for price in prices:
    tax.append(price+0.18*price)
  return tax
```

```python
gst([100,200,250,400])
```

```python
prices=[100,200,250,400]
gst = list(map(lambda price: price+0.18*price, prices))
print(gst)
```

```python
print(list(map(lambda price: price+0.18*price,[100,200,250,400])))
```

#Filter

```python
def even(lst):
  ev=[]
  for elem in lst:
    if elem%2==0:
      ev.append(elem)
  return ev
```

```python
even([1,2,3,4,5,6,7])
```

```python
num =[1,2,3,4,5,6,7]
even=list(filter(lambda x: x%2==0,num))
print(even)
```
