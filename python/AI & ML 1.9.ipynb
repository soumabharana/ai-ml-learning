#Nested Loops

```python
for i in range(3):
  for j in range(2):
    print("i=",i,end=" ")
    print("j=",j)
```

```python
for i in range(3):
    print("i=",i)

for j in range(2):
    print("j=",j)
```

```python
for row in range(3):
  for col in range(4):
    print((col+1)*10,end=" ")
  print()
```

```python
for i in range(6):
  for j in range(i):
    print("*",end=" ")
  print()
```

```python
for i in range(6):
  for j in range(i):
    print(i,end=" ")
  print()
```

```python
for i in range(6):
  for j in range(i):
    print((j+1),end=" ")
  print()
```

```python
k=1
for i in range(6):
  for j in range(i):
    print(k,end=" ")
    k=k+1
  print()
```
