#While Loop

```python
i=1
while i<6:
  print(i)
  i+=1
```

```python
i=1
while i<=7:
  print("Hi")
  i+=1
```

```python
i=0
while i<=10:
  print(i)
  i+=2
```

```python
i=0
while i<=10:
  if i%2==0:
    print(i)
  i+=1
```

```python
i=1
sum =0
while i<=10:
  sum+=i
  i+=1
  print(sum)
```

```python
sum=0
for i in range(1,11):
  sum+=i
  print(sum)
```

```python
i=1
while i<=10:
  print("5 *",i,"=",5*i)
  i+=1
```
