#Variable Naming Rules

```python
# ### Data Types

# 1. **Integer** (`int`)
#    * Examples: `50`, `4`, `-6`, `0`, `22`

# 2. **Float** (`float`)
#    * Decimal values (also referred to as `double` in other languages)
#    * Examples: `5.7`, `-4.5`, `3.6`, `2.0`

# 3. **String** (`str`)
#    * Text data enclosed in quotes
#    * Examples: `"Aditya"`, `"Nagpur"`, `"Aditya Jain"`

# 4. **Boolean** (`bool`)
#    * Logical state values
#    * Examples: `True`, `False`
```

```python
# Quick Data Type Identification Examples
# "4"      --> str   (String because of quotes)
# 5        --> int   (Whole number)
# True     --> bool  (Boolean truth value)
# length = 6.7 --> float (Decimal value)

val1 = "4"
val2 = 5
val3 = True
length = 6.7

print(type(val1))
print(type(val2))
print(type(val3))
print(type(length))
```
