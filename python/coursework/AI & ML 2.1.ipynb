### Note on Floating Point Precision
Computers store numbers in binary (base-2). Just as the fraction $1/3$ cannot be represented exactly in decimal ($0.333...$), the decimal $0.1$ or $1.1$ cannot be represented exactly in binary. This leads to tiny rounding differences like `55000.00000000001` instead of `55000.0`.

To avoid this in real applications (like money), you can use the `round()` function:
```python
print(round(employee['salary'], 2))
```
