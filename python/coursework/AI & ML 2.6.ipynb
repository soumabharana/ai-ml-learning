#Password Validator

```python
password = input("Enter Password: ")
upper = False
lower = False
digit = False

for char in password:
    if char>"A" and char<"Z":
        upper = True
    elif char>"a" and char<'z':
        lower = True
    elif char.isdigit():
        digit = True
if len(password) >= 8 and upper==True and lower==True and digit==True:
    print("Strong Password")
else:
    print("Weak Password")
```

#**Problem 8 : Common Subjects**

```python
**Problem Statement**

Input two sets of subjects.

**Display**

* Common subjects
* Subjects only in first student
* Subjects only in second student
```

```python
A = set()
B = set()

print("Enter subjects for the first student (press Enter without typing to stop):")
while True:
  sub = input("Enter subject name for first student: ")
  if not sub:
    break
  A.add(sub)

print("\nEnter subjects for the second student (press Enter without typing to stop):")
while True:
  sub = input("Enter subject name for second student: ")
  if not sub:
    break
  B.add(sub)
```

```python
# Common subjects
print("Common subjects:", A.intersection(B))

# Subjects only in first student
print("Subjects only in first student:", A.difference(B))

# Subjects only in second student
print("Subjects only in second student:", B.difference(A))
```

#Most Frequent Character

```python
Problem 9 : Most Frequent Character

Problem Statement

Find the character that appears the maximum number of times in a string.

Ignore spaces.
```

```python
s = input("Enter string: ")
freq ={}
for c in s.lower(): # Call the lower() method
  if c == ' ': # Ignore spaces
    continue
  if c not in freq:
    freq[c] = 1
  else: # Increment only if it already exists
    freq[c]+=1
print(freq)
```

```python
max_v =0
for k,v in freq.items():
  if v>max_v:
    max_v=v
    max_k=k
print(max_k,max_v)
```
