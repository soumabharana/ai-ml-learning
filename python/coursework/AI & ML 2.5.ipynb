```python
#Problem 1: Student Grade Analyzer
# Write a Python program that takes marks of 5 subjects as input.
# Your program should:
# Store the marks in a list.
# Calculate the average marks.
# Find the highest and lowest marks.
# Display the grade using:
# Average,Grade
# 90+,A
# 75–89,B
# 60–74,C
# Below 60,D
```

```python
marks = []
for i in range(5):
  m = int(input())
  marks.append(m)
Avg = sum(marks)/len(marks)
print(Avg)
print("Highest Marks: ", max(marks))
print("Lowest Marks: ", min(marks))
if Avg >= 90:
  print("Grade: A")
elif Avg>=75:
  print("Grade:B")
elif Avg>=60:
  print("Grade:C")
else:
  print("Grade:D")
```

```python
# Problem Statement

# Take a sentence as input.

# Display how many times each word appears.

# Example:

# Input: python is fun python is easy

# Output: python - 2 , is - 2 , easy - 1, fun - 1
```

```python
s = input("Enter sentence: ")
words = s.split()
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] += 1  # Increment count if already in dictionary
    else:
        frequency[word] = 1   # Initialize count to 1 if first time seeing it

print(frequency)
# Output: {'python': 2, 'is': 2, 'fun': 1, 'easy': 1}
```

```python
n = input("Enter string: ")
nl = n.lower()
vowel=0
consonent=0
space=0
digits=0
for char in nl:
  if char in "aeiou":
    vowel+=1
  elif char>"a" and char<"z":
    consonent+=1
  elif char in " ":
    space+=1
  elif char>0 and char<9:
    digits+=1
print("Vowels:",vowel)
print("Consonents:",consonent)
print("Space:",space)
print("Digits:",digits)
```

```python
names = []
prices = []

for i in range(5):
    name = input()
    price = int(input())

    names.append(name)
    prices.append(price)

```

```python
total_bill_amount = sum(prices)
mx = prices[0]
mx_prd = ""
mn = prices[0]
mn_prd = ""

for i in range(5):
    if prices[i] > mx:
        mx_prd = names[i]
        mx = prices[i]
    elif prices[i] < mn:
        mn_prd = names[i]
        mn = prices[i]

print(total_bill_amount)
print(mx, mx_prd)
print(mn, mn_prd)
```
