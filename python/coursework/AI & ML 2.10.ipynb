```python
def calc_grade(marks):
    if marks > 90:
        grade = "A"
    elif marks > 75:
        grade = "B"
    elif marks > 60:
        grade = "C"
    else:
        grade = "D"

    return grade
```

```python
for student in students:
  print(student["marks"], calc_grade(student["marks"]))
```

```python
#Create a function to search students by their roll no.
```

```python
def search(roll_no):
  found = False
  for student in students:
    if student["roll"] == roll_no:
      print(student["name"])
      found = True
      return "Record Found!"

  return "Record Not Found"
```

```python
roll = int(input("Enter roll no.: "))
ans = search(roll)
print(ans)
```

```python
# Search Student by name
# ignore case while searching
```

```python
def searches(name):
  found = False
  for student in students:
    if student["name"].lower() == name.lower():
      print(student["name"])
      found = True
      return "Record Found!"

  return "Record Not Found"
```

```python
names = input("Enter name: ")
print(searches(names))
```

```python
def find_topper(students):
  mx_marks = 0
  for student in students:
    if student["marks"] > mx_marks:
      mx_marks = student["marks"]
      topper = student["name"]
  return topper
```

```python
print(find_topper(students))
```

```python
#Count Students in each course
```

```python
def count(students):
  course_count = {}
  for student in students:
    course = student["course"]
    if course in course_count:
      course_count[course] += 1
    else:
      course_count[course] = 1
  return course_count
```

```python
print(count(students))
```

```python
class Students:
  def __init__(self, roll, name, age, course, marks):
    self.roll = roll
    self.name = name
    self.age = age
    self.course = course
    self.marks = marks
  def display_profile(self):
    print("roll:",self.roll, "name:",self.name, "age:",self.age, "course:",self.course, "marks:",self.marks)
```

```python
# Build Complete Student Management Menu:

# 2 Display Students

# 3 Search Student

# 4 Find Topper

# 5 Calculate Average

# 6 Save File

# 7 Exit
```

```python
while True:

    print("\n--- Student Management System Menu ---")
    print("1. Add New Student")
    print("2. Display All Students")
    print("3. Search Student")
    print("4. Find Topper")
    print("5. Calculate Average Marks")
    print("6. Save Student Data")
    print("7. Exit")

    choice_input = input("Enter your choice (1-7): ")

    try:
        choice = int(choice_input)
    except ValueError:
        print("Invalid input. Please enter a number between 1 and 7.")
        continue # Go back to the start of the loop

    if choice == 1:
        # Add New Student (adapted from previous inputs)
        print("\n--- Add New Student ---")
        try:
            roll = int(input("Enter Roll Number: "))
            name = input("Enter Name: ")
            age = int(input("Enter Age: "))
            course = input("Enter Course: ")
            marks = int(input("Enter Marks: "))

            new_student = {
                "roll": roll,
                "name": name,
                "age": age,
                "course": course,
                "marks": marks
            }
            students.append(new_student) # Add to the global students list
            print(f"Student {name} added successfully!")
        except ValueError:
            print("Invalid input. Please enter numbers for Roll, Age, and Marks.")

    elif choice == 2:
        # Display All Students (existing code)
        print("\n--- All Students ---")
        if not students:
            print("No students to display.")
        else:
            for student in students:
                print(student)

    elif choice == 3:
        # Search Student (using existing search and searches functions)
        print("\n--- Search Student ---")
        search_by_choice = input("Search by (R)oll Number or (N)ame? ").lower()
        if search_by_choice == 'r':
            try:
                roll_num = int(input("Enter Roll Number to search: "))
                print(search(roll_num)) # Call existing search by roll function
            except ValueError:
                print("Invalid input for Roll Number. Please enter a number.")
        elif search_by_choice == 'n':
            name_to_find = input("Enter Name to search: ")
            print(searches(name_to_find)) # Call existing searches by name function
        else:
            print("Invalid search choice.")

    elif choice == 4:
        # Find Topper (using existing find_topper function)
        print("\n--- Find Topper ---")
        if not students:
            print("No students to determine topper.")
        else:
            # The find_topper function as previously refined returns the topper's name
            # or a message, so we just print its result.
            print(find_topper(students))

    elif choice == 5:
        # Calculate Average Marks (adapted from previous calculations)
        print("\n--- Calculate Average Marks ---")
        if not students:
            print("No students to calculate average.")
        else:
            total_marks = 0
            for student_item in students:
                total_marks += student_item['marks']
            average = total_marks / len(students)
            print(f"Average marks of all students: {average:.2f}")

    elif choice == 6:
        # Save Student Data (direct implementation)
        print("\n--- Save Student Data ---")
        filename = "students_data.txt"
        try:
            with open(filename, 'w') as f:
                for student_record in students:
                    f.write(str(student_record) + '\n')
            print(f"Student data saved to {filename} successfully!")
        except Exception as e:
            print(f"Error saving data to {filename}: {e}")

    elif choice == 7:
        # Exit
        print("Exiting Student Management System. Goodbye!")
        break

    else:
        print("Invalid choice. Please enter a number between 1 and 7.")
```
