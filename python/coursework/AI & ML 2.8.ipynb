#Class Variable and Instance Variable

```python
class Student:
  college = "IIT Bombay"
  def __init__(self,name):
    self.name = name
s1 = Student("Rahul")
s2 = Student("Raj")
print(s1.name,s1.college)
print(s2.name,s2.college)
```

#Inheritance

```python
# parent class --> the class we are inheriting from --> base class
# child class --> the class that inherits from another class --> derived class
```

```python
class Person:

    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    def show_name(self):
        print(self.first_name, self.last_name)
```

```python
p1 = Person("Raj", "Sharma")
p1.show_name()
```

```python
class Student(Person):
    pass
```

```python
s1 = Student("Raj", "Sharma")
```
